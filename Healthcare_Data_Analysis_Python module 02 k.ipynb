```python

# module 1.....

import pandas as pd

# Sample healthcare dataset
patients = [
    {"Patient_ID": 101, "Age": 45, "BP": 120, "Glucose": 95, "Disease": "Diabetes"},
    {"Patient_ID": 102, "Age": 60, "BP": 145, "Glucose": 180, "Disease": "Diabetes"},
    {"Patient_ID": 103, "Age": 35, "BP": 115, "Glucose": 90, "Disease": "None"},
    {"Patient_ID": 104, "Age": 55, "BP": 150, "Glucose": 160, "Disease": "Hypertension"},
    {"Patient_ID": 105, "Age": 42, "BP": 125, "Glucose": 110, "Disease": "None"},
    {"Patient_ID": 106, "Age": 67, "BP": 160, "Glucose": 190, "Disease": "Diabetes"},
]

df = pd.DataFrame(patients)

# 1. Display patient data
print("HEALTHCARE DATA ANALYSIS")
print(df.to_string(index=False))

# 2. Calculate average age
print("\nAverage Patient Age:", round(df["Age"].mean(), 2))

# 3. Calculate average blood pressure
print("Average Systolic BP:", round(df["BP"].mean(), 2))

# 4. Count disease categories
print("\nDisease Statistics:")
print(df["Disease"].value_counts())

# 5. Filter records for analysis only
print("\nPatients with BP >= 140:")
print(df[df["BP"] >= 140].to_string(index=False))

# 6. Save report
df.to_csv("healthcare_analysis.csv", index=False)
print("\nReport saved as healthcare_analysis.csv")

from google.colab import files
files.download("healthcare_analysis.csv")


# module 2.....

import pandas as pd
import numpy as np
from google.colab import files

# 1. Create a sample healthcare dataset
patients = [
    {"Patient_ID": 101, "Age": 45, "BP": 120, "Glucose": 95, "Disease": "Diabetes"},
    {"Patient_ID": 102, "Age": 60, "BP": 145, "Glucose": 180, "Disease": "Diabetes"},
    {"Patient_ID": 103, "Age": 35, "BP": 115, "Glucose": 90, "Disease": "None"},
    {"Patient_ID": 104, "Age": 55, "BP": 150, "Glucose": 160, "Disease": "Hypertension"},
    {"Patient_ID": 105, "Age": 42, "BP": 125, "Glucose": 110, "Disease": "None"},
    {"Patient_ID": 106, "Age": 67, "BP": 160, "Glucose": 190, "Disease": "Diabetes"},
    {"Patient_ID": 107, "Age": np.nan, "BP": 130, "Glucose": 105, "Disease": "None"},
    {"Patient_ID": 108, "Age": 50, "BP": np.nan, "Glucose": 140, "Disease": "Hypertension"}
]

df = pd.DataFrame(patients)

# 2. Save dataset as CSV
df.to_csv("healthcare_data.csv", index=False)
print("Healthcare dataset created successfully!")

# 3. Import CSV dataset using Pandas
df = pd.read_csv("healthcare_data.csv")

print("\n--- Original Healthcare Dataset ---")
print(df.to_string(index=False))

# 4. Explore the dataset
print("\n--- Dataset Information ---")
print("Shape:", df.shape)
print("\nColumn names:", df.columns.tolist())
print("\nData types:")
print(df.dtypes)

print("\n--- Statistical Summary ---")
print(df.describe())

# 5. Check missing values
print("\n--- Missing Values ---")
print(df.isnull().sum())

# 6. Handle missing values
df["Age"] = df["Age"].fillna(df["Age"].median())
df["BP"] = df["BP"].fillna(df["BP"].median())

print("\n--- Dataset After Cleaning ---")
print(df.to_string(index=False))
print("\nMissing values after cleaning:")
print(df.isnull().sum())

# 7. Check and remove duplicate records
print("\nDuplicate records:", df.duplicated().sum())
df = df.drop_duplicates()

# 8. NumPy analysis
print("\n--- NumPy Analysis ---")
print("Average Age:", round(np.mean(df["Age"]), 2))
print("Average BP:", round(np.mean(df["BP"]), 2))
print("Average Glucose:", round(np.mean(df["Glucose"]), 2))

# 9. Disease category analysis
print("\n--- Disease Statistics ---")
print(df["Disease"].value_counts())

# 10. Filter patients with BP >= 140
print("\n--- Patients with BP >= 140 ---")
print(df[df["BP"] >= 140].to_string(index=False))

# 11. Save cleaned dataset
df.to_csv("healthcare_analysis_cleaned.csv", index=False)
print("\nCleaned dataset saved successfully!")

# 12. Download cleaned CSV file
files.download("healthcare_analysis_cleaned.csv")





```
